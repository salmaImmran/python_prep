Two Sum (Hash Map Pattern)

```python
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        need = target - num
        if need in seen:
            return [seen[need], i]
        seen[num] = i
    return []
    return []
```

Mutable Default Argument Trap

```python
def add_item(item, bucket=None):
    if bucket is None:
        bucket = []
        bucket.append(item)
        return bucket
```

Yeild and Return 

```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for num in count_up_to(3):
    print(num)
```

Decorator in Python

```python
def log_call(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}...")
        return func(*args, **kwargs)
    return wrapper

@log_call
def add(a, b):
    return a + b
print(add(2, 3))
```

*args and **kwargs

```python
def greet(*names, **details):
    for name in names:
        print(f"Hello, {name}!")
        print("Extra info:", details)
greet("Asha", "Ravi", role="Intern", track="Python")
```

Difference between '==' and 'is'

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b) 
print(a is b)
```

Mutable and Immutable

```python
# Immutable (str / int)
x = 10
y = x
x = x + 5 # Creates a new object 15; y still points to 10
# Mutable (list)
a = [1, 2]
b = a # Both a and b point to the same list
b.append(3)
print(a) # Output: [1, 2, 3]
```

Valid Palindrome (Two Pointers Pattern)

```python
def is_palindrome(s):
    left, right = 0, len(s) - 1
    while left < right:
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        if s[left].lower() != s[right].lower():
            return False
            left += 1
            right -= 1
    return True
```

Valid Anagram (Frequency Counter Pattern)

```python
def is_anagram(s, t):
    if len(s) != len(t):
        return False
        counts = {}
    for ch in s:
        counts[ch] = counts.get(ch, 0) + 1
    for ch in t:
        if counts.get(ch, 0) == 0:
            return False
        counts[ch] -= 1
    return True
```
