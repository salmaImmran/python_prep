Linear Search

```python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1
list_of_numbers = [13, 4, 7, 9, 10]
target_number = 9
result = linear_search(list_of_numbers, target_number)
print(f"Target number {target_number} found at index: {result}" if result != -1 
      else f"Target number {target_number} not found in the list.")
```

Binary Search

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2  
        
        print(f"Searching in the range: {arr[low:high+1]}, mid element: {arr[mid]}")  # Debugging line

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1
numbers = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
target = 56
result = binary_search(numbers, target)
print(f"Target number {target} found at index: {result}" if result != -1
      else f"Target number {target} not found in the list.")
```

Merge sort-Divide and Conquer

```python
def merge_sort(arr):
    print(f"Sorting array: {arr}")  # Debugging line
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    sorted_list = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            sorted_list.append(left[i])
            i += 1
        else:
            sorted_list.append(right[j])
            j += 1

    sorted_list.extend(left[i:])
    sorted_list.extend(right[j:])
    print(f"Merging: {left} and {right} -> {sorted_list}")  # Debugging line

    return sorted_list
test_arr = [4, 2, 7, 1, 3]
print(merge_sort(test_arr))
```
